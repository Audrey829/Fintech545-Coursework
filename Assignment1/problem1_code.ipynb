```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# --------------------------------------------------
# Problem 1: Read the sample
# --------------------------------------------------

# Load data
data = pd.read_csv("problem1.csv")

# problem1.csv contains one variable, x
x = data["x"].dropna()

n = len(x)

# --------------------------------------------------
# First four sample moments
# --------------------------------------------------

# 1. Sample mean
mean_x = np.mean(x)

# 2. Sample variance
variance_x = np.var(x, ddof=1)

# 3. Sample skewness
skewness_x = skew(x, bias=False)

# 4. Sample excess kurtosis
# fisher=True means we report EXCESS kurtosis:
# bias=False applies the finite-sample bias correction.
excess_kurtosis_x = kurtosis(x, fisher=True, bias=False)

# --------------------------------------------------
# Print results
# --------------------------------------------------

print("Problem 1 - First Four Sample Moments")
print("--------------------------------------")
print(f"Number of observations: {n}")
print(f"Mean:               {mean_x:.6f}")
print(f"Variance:           {variance_x:.6f}")
print(f"Skewness:           {skewness_x:.6f}")
print(f"Excess Kurtosis:    {excess_kurtosis_x:.6f}")
```

```python
from scipy.stats import norm

# --------------------------------------------------
# Problem 1(b): Fit a Normal by matching
# the sample mean and sample variance
# --------------------------------------------------

std_x = np.sqrt(variance_x)

# The fitted Normal distribution is N(mean_x, variance_x).
# Find its 1% quantile.
normal_q01 = norm.ppf(0.01, loc=mean_x, scale=std_x)

# Count the actual sample observations below
# the fitted Normal's 1% quantile.
actual_below = np.sum(x < normal_q01)

# Under the fitted Normal, 1% of the observations
# are expected to fall below its 1% quantile.
expected_below = 0.01 * n

print("\nProblem 1(b) - Normal Fit")
print("-------------------------")
print(f"Fitted mean:              {mean_x:.6f}")
print(f"Fitted standard deviation:{std_x:.6f}")
print(f"Normal 1% quantile:       {normal_q01:.6f}")
print(f"Actual observations below:{actual_below}")
print(f"Expected observations:    {expected_below:.1f}")
```
